## UN Comtrade에서 무역 데이터 불러오기
- UN Comtrade API 동작 확인용 노트북

#### API 호출에 쓰이는 param
- typeCode = "C": 상품 / "S": 서비스
- freqCode = "A": 연간 데이터 / "M": 월별 데이터
- period = "2022": 연간 / "202205": 월, (여려 연도나 월일 경우 $\Rightarrow$ "2021,2022,2023")
- reporterCode: 기준이 되는 나라의 코드
- partnerCode: 상대 나라의 코드
- clCode = "HS" (품목 분류 기준, 건들 필요 X)
- cmdCode = "72": 철강 / "TOTAL": 전체 품목... (품목 코드)
- flowCode = "X": 수출 / "M": 수입
- partner2Code: 제2 중간국
- customsCode: 통관방식
- motCode: 운송수단

```python
import pandas as pd
import comtradeapicall
from src.config import get_comtrade_api_key
```

```python
API_KEY = get_comtrade_api_key() # 프로젝트 내 src/config.py에서 API 키 가져오는 방식으로 수정 (하드코딩x)
```

#### 베트남 -> 한국으로 철강 수출 데이터 2020~2024년 연간으로 조회
-  중국산 철강이 베트남을 경유해 한국에 수출되는 무역 우회 패턴 탐지 (전형적인 검증 대상 중 하나)

```python
df = comtradeapicall.getFinalData(
    API_KEY,
    typeCode='C',                     # 상품
    freqCode='A',                     # 연간
    clCode='HS',                      # HS 분류
    period='2020,2021,2022,2023,2024',
    reporterCode='704',               # 베트남(기준)
    cmdCode='72',                     # 철강
    flowCode='X',                     # 수출
    partnerCode='410',                # 한국 => [베트남 -> 한국]
    partner2Code=None,
    customsCode=None,
    motCode=None,
    maxRecords=2500,
    format_output='JSON',
    aggregateBy=None,
    breakdownMode='classic',
    countOnly=None,
    includeDesc=True
)
```
